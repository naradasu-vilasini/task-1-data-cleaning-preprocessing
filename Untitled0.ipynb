```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Titanic-Dataset.csv")
print(df.shape)
df.info()
print(df.isnull().sum())
df.describe()
```

```python
# Age is numeric, so fill with the median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Embarked is text, so fill with the most common value (mode)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Cabin is mostly empty, and these columns are not useful, so drop them
df = df.drop(columns=['Cabin', 'PassengerId', 'Name', 'Ticket'])

print(df.isnull().sum())  # should now be all 0
```

```python
# Sex has only 2 values, so use label encoding
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Embarked has 3 values, so use one-hot encoding
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True, dtype=int)

df.head()
```

```python
cols = ['Age', 'Fare', 'SibSp', 'Parch']
plt.figure(figsize=(12, 4))
for i, c in enumerate(cols):
    plt.subplot(1, 4, i + 1)
    sns.boxplot(y=df[c])
    plt.title(c)
plt.tight_layout()
plt.savefig("boxplots_before.png")
plt.show()
```

```python
def remove_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return data[(data[column] >= lower) & (data[column] <= upper)]

print("Before:", df.shape)
for c in ['Age', 'Fare']:
    df = remove_outliers(df, c)
print("After:", df.shape)

plt.figure(figsize=(8, 4))
for i, c in enumerate(['Age', 'Fare']):
    plt.subplot(1, 2, i + 1)
    sns.boxplot(y=df[c])
    plt.title(c + " (after)")
plt.tight_layout()
plt.savefig("boxplots_after.png")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])

df.head()
```

```python
df.to_csv("titanic_cleaned.csv", index=False)
```
